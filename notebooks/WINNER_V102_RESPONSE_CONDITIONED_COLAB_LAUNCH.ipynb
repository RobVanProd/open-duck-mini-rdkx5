```python
import hashlib
import subprocess
import sys
from pathlib import Path

package = Path('/content/winner-v102-response-conditioned-hosted-20260722.tar.gz')
launcher = Path('/content/launch_winner_v102_response_conditioned_colab.py')
def sha256(path):
    digest = hashlib.sha256()
    with path.open('rb') as stream:
        for block in iter(lambda: stream.read(1024 * 1024), b''):
            digest.update(block)
    return digest.hexdigest()
assert package.stat().st_size == 59_176_688
assert sha256(package) == 'cbb6dd1ef5c68e013cbc8e0a29bdbb014cff476064f2ac763aedfeeba6b28017'
assert sha256(launcher) == '24d942f85b2e8bdeb43493f27c78c111ea0e1e4e96c33f21bb30a55674de1818'
command = [sys.executable, str(launcher), '--package', str(package), '--hosted-gpu-authorized']
completed = subprocess.run(command, text=True, check=False)
assert completed.returncode == 0, completed.returncode
```
